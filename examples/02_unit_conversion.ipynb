# Unit Conversion in Physities

This notebook demonstrates unit conversion capabilities.

```python
from physities.src.unit import (
    Meter, Kilometer, Mile, Foot, Inch,
    Second, Minute, Hour,
    Kilogram, Gram, Pound,
)
```

## Length Conversions

```python
# Convert between metric units
km = Kilometer(10)
m = km.convert(Meter)
print(f"{km.value} km = {m.value} m")

# Convert to imperial
miles = km.convert(Mile)
print(f"{km.value} km = {miles.value:.2f} miles")

# Convert imperial to metric
ft = Foot(100)
m = ft.convert(Meter)
print(f"{ft.value} feet = {m.value:.2f} meters")
```

## Speed Conversions

```python
# Define speed unit types
Ms = Meter / Second
Kh = Kilometer / Hour
Mph = Mile / Hour

# Convert m/s to km/h
speed = Ms(30)
print(f"{speed.value} m/s = {speed.convert(Kh).value} km/h")

# Convert km/h to mph
speed_kmh = Kh(100)
print(f"{speed_kmh.value} km/h = {speed_kmh.convert(Mph).value:.2f} mph")
```

## Mass Conversions

```python
# Metric to imperial
kg = Kilogram(70)
lb = kg.convert(Pound)
print(f"{kg.value} kg = {lb.value:.1f} lbs")

# Grams to kilograms
g = Gram(500)
kg = g.convert(Kilogram)
print(f"{g.value} g = {kg.value} kg")
```

## Converting to SI Base Units

```python
# to_si() converts any unit to SI base units
km = Kilometer(5)
si = km.to_si()
print(f"{km.value} km -> {si.value} m (SI)")

lb = Pound(10)
si = lb.to_si()
print(f"{lb.value} lb -> {si.value:.2f} kg (SI)")
```
