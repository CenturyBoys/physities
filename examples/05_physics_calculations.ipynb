# Physics Calculations with Physities

This notebook demonstrates common physics calculations.

```python
from physities.src.unit import Meter, Second, Kilogram, Kilometer, Hour
```

## Kinematics

```python
# Define unit types
Velocity = Meter / Second
Acceleration = Meter / (Second ** 2)

# Free fall from rest
g = Acceleration(9.8)  # gravity
t = Second(5)  # time
v0 = Velocity(0)  # initial velocity

# v = v0 + g*t
v = v0 + g * t
print(f"Velocity after {t.value}s: {v}")

# s = v0*t + 0.5*g*t^2
s = v0 * t + (g * t * t) * 0.5
print(f"Distance fallen: {s}")
```

## Force and Work

```python
# Newton's second law: F = ma
m = Kilogram(10)
a = Acceleration(5)
F = m * a
print(f"Force: {F}")

# Work: W = F * d
d = Meter(20)
W = F * d
print(f"Work: {W}")

# Power: P = W / t
t = Second(4)
P = W / t
print(f"Power: {P}")
```

## Kinetic Energy

```python
# E = 0.5 * m * v^2
m = Kilogram(1000)  # 1 ton car
v = (Kilometer / Hour)(100)  # 100 km/h

# Convert to m/s for calculation
v_ms = v.convert(Meter / Second)
print(f"Speed: {v_ms.value:.2f} m/s")

# Kinetic energy
E = m * v_ms * v_ms * 0.5
print(f"Kinetic energy: {E}")
```

## Momentum

```python
# p = m * v
m1 = Kilogram(5)
v1 = (Meter / Second)(10)
p1 = m1 * v1
print(f"Momentum of object 1: {p1}")

m2 = Kilogram(3)
v2 = (Meter / Second)(-8)  # Opposite direction
p2 = m2 * v2
print(f"Momentum of object 2: {p2}")

# Total momentum (conservation)
p_total = p1 + p2
print(f"Total momentum: {p_total}")
```
