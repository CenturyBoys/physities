# Dimensional Analysis with Physities

This notebook shows how to use dimensional analysis to verify physics equations.

```python
from physities.src.unit import Meter, Second, Kilogram
from physities.src.dimension import Dimension
```

## Understanding Dimensions

Every physical quantity has a dimension expressed in terms of 7 SI base dimensions.

```python
# Create base dimensions
length = Dimension.new_length()
mass = Dimension.new_mass()
time = Dimension.new_time()

print(f"Length dimension: L={length.length}")
print(f"Mass dimension: M={mass.mass}")
print(f"Time dimension: T={time.time}")
```

## Derived Dimensions

```python
# Velocity = Length / Time
Velocity = Meter / Second
print(f"Velocity: L={Velocity.scale.dimension.length}, T={Velocity.scale.dimension.time}")

# Acceleration = Length / Time^2
Acceleration = Meter / (Second ** 2)
print(f"Acceleration: L={Acceleration.scale.dimension.length}, T={Acceleration.scale.dimension.time}")

# Force = Mass * Acceleration
Newton = Kilogram * Acceleration
print(f"Force: L={Newton.scale.dimension.length}, M={Newton.scale.dimension.mass}, T={Newton.scale.dimension.time}")
```

## Verifying Equations

Dimensional analysis can verify that equations are dimensionally consistent.

```python
# Kinetic Energy: E = 0.5 * m * v^2
m = Kilogram(10)
v = (Meter / Second)(5)

# Calculate kinetic energy
E = m * v * v * 0.5

# Check dimension (should be M*L^2/T^2 = Energy)
print(f"Energy dimension: L={E.scale.dimension.length}, M={E.scale.dimension.mass}, T={E.scale.dimension.time}")
print(f"Kinetic energy: {E}")
```

## Dimensionless Quantities

```python
# Ratio of same-dimension quantities is dimensionless
d1 = Meter(100)
d2 = Meter(50)
ratio = d1 / d2

print(f"Ratio: {ratio.value}")
print(f"Is dimensionless: {ratio.scale.is_dimensionless}")
```
