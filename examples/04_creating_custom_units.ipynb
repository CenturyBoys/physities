# Creating Custom Units

This notebook shows different ways to create custom unit types.

```python
from physities.src.unit import Meter, Second, Kilogram, Unit, Mile, Hour
from physities.src.scale import Scale
from physities.src.dimension import Dimension
```

## Method 1: Scaling Existing Units

The simplest way to create a custom unit is to scale an existing one.

```python
# Create a furlong (1/8 mile = 201.168 meters)
Furlong = Meter * 201.168

# Use it
distance = Furlong(10)
print(f"10 furlongs = {distance.convert(Meter).value:.2f} meters")

# Create a nautical mile (1852 meters)
NauticalMile = Meter * 1852
nm = NauticalMile(5)
print(f"5 nautical miles = {nm.convert(Meter).value} meters")
```

## Method 2: Combining Units

Create composite units by combining existing units.

```python
# Knot = nautical miles per hour
NauticalMile = Meter * 1852
Knot = NauticalMile / Hour

speed = Knot(10)
Ms = Meter / Second
print(f"10 knots = {speed.convert(Ms).value:.2f} m/s")

# Acceleration due to gravity
G = (Meter / (Second ** 2))(9.80665)
print(f"Standard gravity: {G}")
```

## Method 3: Using Scale Directly

For precise control, create a Scale directly.

```python
# Create an Angstrom (10^-10 meters)
angstrom_scale = Scale.new(
    dimension=Dimension.new_length(),
    from_base_scale_conversions=(1e-10, 1, 1, 1, 1, 1, 1),
)

class Angstrom(Unit):
    scale = angstrom_scale
    value = None

# Green light wavelength
wavelength = Angstrom(5500)
nm = wavelength.to_si().value * 1e9
print(f"5500 Angstroms = {nm:.1f} nm")
```

## Method 4: Multi-dimensional Custom Units

```python
# Create a custom energy unit: calorie (4.184 joules)
# Joule = kg * m^2 / s^2
Joule = Kilogram * Meter * Meter / (Second ** 2)
Calorie = Joule * 4.184

# Use it
energy = Calorie(100)
joules = energy.convert(Joule)
print(f"100 calories = {joules.value} joules")
```
