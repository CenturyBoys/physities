# Getting Started with Physities

This notebook introduces the basics of using Physities for physical quantity calculations.

```python
# Import basic units
from physities.src.unit import Meter, Second, Kilogram, Kilometer, Hour
```

## Creating Quantities

Create physical quantities by instantiating unit classes with a numeric value:

```python
# Create some basic quantities
distance = Meter(100)
time = Second(10)
mass = Kilogram(5)

print(f"Distance: {distance}")
print(f"Time: {time}")
print(f"Mass: {mass}")
```

## Composite Units

Create composite unit types using arithmetic operators:

```python
# Velocity = distance / time
Velocity = Meter / Second

# Create a velocity value
speed = Velocity(10)
print(f"Speed: {speed}")

# Or calculate from other quantities
calculated_speed = distance / time
print(f"Calculated speed: {calculated_speed}")
```

## Arithmetic Operations

Perform arithmetic with proper dimensional handling:

```python
# Addition (same units only)
d1 = Meter(50)
d2 = Meter(30)
total = d1 + d2
print(f"{d1} + {d2} = {total}")

# Multiplication creates new dimensions
area = Meter(10) * Meter(5)
print(f"Area: {area}")

# Scalar operations
doubled = d1 * 2
print(f"{d1} * 2 = {doubled}")
```

## Unit Conversion

Convert between compatible units:

```python
# Convert meters to kilometers
m = Meter(5000)
km = m.convert(Kilometer)
print(f"{m.value} m = {km.value} km")

# Convert speed units
Ms = Meter / Second
Kh = Kilometer / Hour

speed_ms = Ms(10)
speed_kmh = speed_ms.convert(Kh)
print(f"{speed_ms.value} m/s = {speed_kmh.value} km/h")
```

## Unit Equality

Units representing the same physical quantity are equal:

```python
# Same physical quantity in different units
m = Meter(1000)
km = Kilometer(1)

print(f"Meter(1000) == Kilometer(1): {m == km}")
```
